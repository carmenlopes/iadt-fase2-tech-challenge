# Tabela Comparativa entre os modelos Regressão Logística X Árvore de Decisão

```python
import pandas as pd
comparacao_modelos = pd.DataFrame({
    "Métrica": [
        "Accuracy",
        "Recall",
        "Precision",
        "F1-score",
        "Falsos Positivos",
        "Falsos Negativos"
    ],
    "Regressão Logística": [
        "96,49%",
        "92,86%",
        "97,50%",
        "95,12%",
        1,
        3
    ],
    "Árvore de Decisão": [
        "93,86%",
        "88,10%",
        "94,87%",
        "91,36%",
        2,
        5
    ]
})

display(comparacao_modelos)
```

# Comparação entre os modelos

Analisando o desempenho dos dois modelos, podemos concluir que a Regressão Logística apresentou um desempenho melhor em todas as métricas, inclusive no Recall, utilizado como referência devido à necessidade do experimento de obter a maior quantidade possível de casos malignos previstos corretamente. Enquanto a Árvore de Decisão previu **7 casos erroneamente, sendo 5 de falsos negativos**, em contrapartida a **Regressão Logística previu incorretamente 4 casos, sendo 3 falsos negativos.**

Com base nos resultados, a Regressão Logística também apresentou melhor F1-score, de **95,12%** contra **91,36%** da Árvore de Decisão, indicando um melhor equilíbrio entre a Precisão e o Recall. A acurácia também foi superior, com **96,49%** na Regressão Logística contra **93,86%** na Árvore de Decisão, uma diferença de <u>**2,63**</u> pontos percentuais.

Embora o desempenho da Regressão tenha sido superior, os casos classificados incorretamente ainda podem ser considerados preocupantes, principalmente os 3 casos que eram malignos, mas foram previstos como benignos. Em um cenário real, caso o resultado do modelo fosse utilizado sem uma segunda avaliação de um profissional de saúde, esses falsos negativos poderiam representar um risco para as pacientes.

Dessa forma, apesar do bom desempenho, eu não considero que o modelo está apto para ser utilizado em um cenário prático da vida real. Devido à existência de falsos negativos e às limitações do experimento, como a quantidade relativamente baixa de registros do dataset, a previsão realizada pelo modelo não deve ser considerada um diagnóstico definitivo. Com mais registros e validações, futuramente o modelo poderia ser refinado e utilizado como uma ferramenta de apoio à triagem de casos, mas posteriormente todos os casos deveriam ser analisados por um profissional de saúde. A decisão final deve ser tomada por um profissional competente da área médica.
